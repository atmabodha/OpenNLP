```python
from newspaper import Article
```

```python
def news(url):
    article = Article(url, language="en")
    try:
        article.download()
    except:
        print("Download Failed")
        return 'error'
    article.parse()
    text = article.text
    if (len(text.split()))<100:
        return 'error'
    return text
```
