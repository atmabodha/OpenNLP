```python
from moviepy import VideoFileClip, concatenate_videoclips
import cv2
```

```python
def blur_video_segment(input_file, output_file, start_time, end_time):
    # 1. Load the original video
    video = VideoFileClip(input_file)
    
    # 2. Slice the video into three distinct parts (v2 uses 'subclipped')
    clip_before = video.subclipped(0, start_time)
    clip_to_blur = video.subclipped(start_time, end_time)
    clip_after = video.subclipped(end_time, video.duration)
    
    # 3. Define a function to blur individual frames using OpenCV
    def blur_frame(frame):
        # (71, 71) is the kernel size. Higher numbers = heavier blur. 
        # Note: These numbers must always be odd.
        return cv2.GaussianBlur(frame, (71, 71), 0)
        
    # 4. Apply the blur function to the middle clip (v2 uses 'image_transform')
    blurred_clip = clip_to_blur.image_transform(blur_frame)
    
    # 5. Stitch the three clips back together in order
    final_video = concatenate_videoclips([clip_before, blurred_clip, clip_after])
    
    # 6. Export the final video
    final_video.write_videofile(
        output_file, 
        codec="libx264", 
        audio_codec="aac"
    )
    
    # Close the clips to free up memory
    video.close()
    final_video.close()
```

```python

# --- Run the function ---
blur_video_segment("clean_video.mp4", "blur_video.mp4", start_time=13, end_time=19)
```
